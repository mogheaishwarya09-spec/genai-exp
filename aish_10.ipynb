```python
import torch
from diffusers import DiffusionPipeline

model_id = "damo-vilab/text-to-video-ms-1.7b"

pipe = DiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    variant="fp16"
)

pipe = pipe.to("cuda")

print("Model loaded successfully!")
```

```python
prompt = "A small boat sailing on a calm lake at sunrise, cinematic view"

result = pipe(prompt, num_frames=16, num_inference_steps=25)
frames = result.frames[0]

print("Video generated successfully!")
```

```python
import imageio.v2 as imageio
from IPython.display import Video, display

imageio.mimsave(
    "generated_video.mp4",
    frames,
    fps=8
)

display(Video("generated_video.mp4", embed=True))

print("Video saved successfully!")
```

```python
prompts = [
    "A waterfall flowing through a green forest, cinematic view",
    "A small boat sailing on a calm lake at sunrise",
    "A robot walking in a futuristic city, science fiction"
]

import imageio.v2 as imageio
from IPython.display import Video, display

for i, prompt in enumerate(prompts, start=1):
    print(f"Generating video {i}: {prompt}")

    result = pipe(
        prompt,
        num_frames=16,
        num_inference_steps=25
    )

    frames = result.frames[0]
    filename = f"video_{i}.mp4"

    imageio.mimsave(filename, frames, fps=8)

    display(Video(filename, embed=True))
    print(f"Saved: {filename}")
```
