```python
import tensorflow as tf
import numpy as np

print("TensorFlow Version:", tf.__version__)
```

```python
scalar = tf.constant(10)
vector = tf.constant([1, 2, 3])
matrix = tf.constant([[1, 2], [3, 4]])
tensor3d = tf.constant([[[1, 2], [3, 4]]])

print("Scalar:", scalar)
print("Vector:", vector)
print("Matrix:\n", matrix)
print("3D Tensor:\n", tensor3d)
```

```python
print("Shape:", matrix.shape)
print("Rank:", tf.rank(matrix).numpy())
print("Data Type:", matrix.dtype)
```

```python
a = tf.constant([10, 20, 30])
b = tf.constant([2, 4, 5])

print("Addition:", tf.add(a, b).numpy())
print("Subtraction:", tf.subtract(a, b).numpy())
print("Multiplication:", tf.multiply(a, b).numpy())
print("Division:", tf.divide(a, b).numpy())
```

```python
x = tf.constant([4.0, 9.0, 16.0])

print("Square:", tf.square(x).numpy())
print("Square Root:", tf.sqrt(x).numpy())
print("Mean:", tf.reduce_mean(x).numpy())
print("Maximum:", tf.reduce_max(x).numpy())
```

```python
x = tf.constant([1, 2, 3, 4, 5, 6])

reshaped = tf.reshape(x, [2, 3])

print("Original Tensor:", x.numpy())
print("Reshaped Tensor:\n", reshaped.numpy())
```

```python
tensor = tf.constant([10, 20, 30])

numpy_array = tensor.numpy()

print("TensorFlow Tensor:", tensor)
print("NumPy Array:", numpy_array)
print("Array Type:", type(numpy_array))
```
